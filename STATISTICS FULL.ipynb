## 1)The average height of all residents in a city is 168cm with a population std = 3.9. A doctor believes the mean is different. He measured the height of 36 individuals and found the average to be 169.5cm 
### (a). State Null and Alternative hypothesis
### (b) At a 95% ,CI is there enough evidence to reject the Null Hypothesis.

```python
# GIVEN DATA
pop_mean=168
pop_std_dev=3.9
n=36
x_bar=169.5
```

```python
# H0: The mean is not different.
# Ha: The mean is different.
```

```python
import numpy as np
z_score=(x_bar-pop_mean)/(pop_std_dev/np.sqrt(n))
z_score
```

```python
import scipy.stats as stats
z_critical=stats.norm.ppf(0.975)
z_critical 
```

# 2)Scores on an exam are normally distributed with a population standard deviation=5.6. a random sample of 40 scores on the exam has a mean of 32.
# Estimate the population mean with
# a)80% confidence 
# b)90% confidence
# c)98 confidences

```python
pop_mean_upper_90= (x_bar+z_critical_90*(std_dev/np.sqrt(n)))
pop_mean_upper_90
```

# 5.A statistics teacher wants to compare his two classes to see if they performed any differently on the tests he gave that semester. Class A had 25 students with an average score of 70, standard deviation 15. Class B had 20 students with an average score of 74 , standard deviation 25. Using alpha 0.05 did these two classes perform differently on the tests?.

```python
# GIVEN
n1=25
x1_bar=70
s1=15
n2=20
x2_bar=74
s2=25
# df1=n1-1=25-1=24
# df2=n2-1=20-1=19
df1=24
df2=19
```

```python
sp_square=(((s1**2) * df1) + ((s2**2) * df2)) / (df1 + df2)
sp_square
```

```python
t_score=(x1_bar-x2_bar)/np.sqrt((sp_square/n1)+(sp_square/n2))
t_score
```

```python
t_critical=stats.t.ppf(0.025,43)
t_critical
```

# 8)Anova Test
# a= [2,3,7,2,6]
# b=[10,8,7,5,10]
# c=[10,13,14,13,15]

```python
a= [2,3,7,2,6]
b=[10,8,7,5,10]
c=[10,13,14,13,15]
# GIVEN DATA
n=5
df1=2
df2=12
```

```python
# VARIENCE BTE THE GRP
mean_a=np.mean(a)
mean_b=np.mean(b)
mean_c=np.mean(c)
overall_mean=(mean_a+mean_b+mean_c)/3
vbg=n*((overall_mean-mean_a)**2)+n*((overall_mean-mean_b)**2)+n*((overall_mean-mean_c)**2)
vbg
```

```python
# VARIENCE WITH IN THE GRP
var_a=sum((a-mean_a)**2)
var_b=sum((b-mean_b)**2)
var_c=sum((c-mean_c)**2)
vwg=var_a+var_b+var_c
vwg
```

```python
f_score=(vbg/df1)/(vwg/df2)
f_score
```

```python
f_critical=stats.f.ppf(0.95,2,12)
f_critical
```
