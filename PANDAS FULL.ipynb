```python
# concat()---> used to combine 2 or more dataframe
```

```python
employee1 = {
    'Name': ['Arun', 'Meera', 'Rahul'],
    'Age': [24, 29, 32],
    'Department': ['IT', 'HR', 'Finance']
}

employee2 = {
    'Name': ['Anjali', 'Vishnu', 'Sneha'],
    'Age': [26, 35, 28],
    'Department': ['IT', 'Sales', 'HR']
}
```

```python
dfa=pd.DataFrame(employee1)
dfa
```

```python
dfb=pd.DataFrame(employee2)
dfb
```

```python
# in pandas row---> axis=0
#            column-> axis=1
# Combine dfa and dfb row-wise.
pd.concat([dfa,dfb],axis=0)
```

```python
# Combine dfa and dfb side by side (column-wise).
pd.concat([dfa,dfb],axis=1)
```

```python
# Combine dfa and dfb row-wise and reset the index.
pd.concat([dfa,dfb],axis=0,ignore_index=True)
```

```python
# groupby()---> groups data based on a column.
# Find the total salary for each department.
df5.groupby('Department')['Salary'].sum()
```

```python
# Find the average salary for each department.
df5.groupby('Department')['Salary'].sum()
```

```python
# Find the number of employees in each department.
df5.groupby('Department')['Salary'].count()
```

```python
# Find the lowest salary in each department.
df5.groupby('Department')['Salary'].min()
```

```python
# Find the highest salary in each department.
df5.groupby('Department')['Salary'].max()
```

```python
# Find the total, average, minimum,maximum ,count salary for each department.
df5.groupby('Department')['Salary'].agg(['sum','mean','min','max','count'])
```

```python
# pivot_table()---> create a summary table like excel pivot table.
# The average salary for each Department, separated by City. create a pivot table
df5.pivot_table(index='Department',columns='City',values='Salary',aggfunc='mean')
```

```python
#Create a pivot table showing the maximum Performance_Score for each Department, separated by City.
df5.pivot_table(index='Department',columns='City',values='Performance_Score',aggfunc='max')
```

```python
# Create a pivot table showing the total Salary for each Department, separated by City.
df5.pivot_table(index='Department',columns='City',values='Salary',aggfunc='sum')
```

```python
# crosstab()---> create frequency/count table between categorical column.
# Create a crosstab showing the number of employees in each Department, separated by City, and include the Total.
pd.crosstab(
    index=df5['Department'],
    columns=df5['City'],
    margins=True,
    margins_name='Total'
)
```

```python
# Using df5, create a crosstab that shows the number of employees in each City based on their Department, and include a Total row and column.
pd.crosstab(index=df5['Department'],
            columns=df5['City'],
            margins=True,
            margins_name='Total')
```

```python
# Create a crosstab showing the number of employees in each City based on their Performance Score.
pd.crosstab(index=df5['Performance_Score'],
            columns=df5['City'])
```

```python
# Create a crosstab showing the number of employees in each Department based on their Experience_Years.
pd.crosstab(index=df5['Experience_Years'],
            columns=df5['Department'])
```
