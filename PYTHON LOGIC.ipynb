# PRIME NUMBER OR NOT

```python
# Write a Python program to check whether .num = 10 is prime or not
num=10
count=0
for i in range (2,num):
    if num%i==0:
        count=count+1
if count==0:
    print('Prime')
else:
    print(num,'Not prime')
    
```
