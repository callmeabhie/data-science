# cosine similarity

## 1) item-based

```python
# we can see the rating values having decimal. for doing this we need whole number format so we need to convert to pivot_table
# here item is movie. we need set item which means index='movie',columns='user_id',value=rating
df1=df.pivot_table(index='movie',columns='userId',values='rating')
df1.fillna(0,axis=1,inplace=True)
df1.head()
```

```python
df1.index
```

```python
# cosine similarity
cos_sim=cosine_similarity(df1)
cos_sim
```

```python
def recommended_movie(similar_movie):
    if similar_movie in df1.index:
        index = np.where(similar_movie == df1.index)[0][0]
        similar = sorted(list(enumerate(cos_sim[index])),reverse=True,key=lambda x: x[1])[1:6]
        print("Recommended movie of ", similar_movie)
        for i in similar:
            print(df1.index[i[0]])
    else:
        print("Movie is not in the list")
```

```python
recommended_movie('titanic')
```

## 2) user-based

```python
df2=df.pivot_table(index='userId',columns='movie',values='rating')
df2.fillna(0,axis=1,inplace=True)
df2.head()
```

```python
# calculate the cosine_similarity and convert into Dataframe and stored in df3 
df3=pd.DataFrame(cosine_similarity(df2),index=df.userId.unique(),columns=df.userId.unique())
df3[df3[3]>0.8][3].sort_values(ascending=False)[1:6]
```

# 2)Eucledian distance

## item_based

```python
df1=df.pivot_table(index='movie',columns='userId',values='rating')
df1.fillna(0,axis=1,inplace=True)
df1.head()
```

```python
euclidean_distance=euclidean_distances(df1)
```

```python
def recommended_movie(similar_movie):
    if similar_movie in df1.index:
        index = np.where(similar_movie == df1.index)[0][0]
        similar = sorted(list(enumerate(euclidean_distance[index])),reverse=False,key=lambda x: x[1])[1:6]
        print("Recommended movie of ", similar_movie)
        print("*"*30)
        for i in similar:
            print(df1.index[i[0]])
    else:
        print("Movie is not in the list")
```

```python
recommended_movie('Jumanji (1995)')
```

## user_based

```python
df2=df.pivot_table(index='userId',columns='movie',values='rating')
df2.fillna(0,axis=1,inplace=True)
df2.head()
```

```python
df3=pd.DataFrame(euclidean_distance(df2),index=df.userId.unique(),columns=df.userId.unique())
df3[df3[10]<1][10].sort_values()[1:6]
```
