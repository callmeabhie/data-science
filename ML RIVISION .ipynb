```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.linear_model import LinearRegression,Lasso,Ridge,ElasticNet
```

```python
df=pd.read_csv('sl1.csv')
df.head()
```

```python
df.info()
```

```python
# set target and features
target=df[['Price']]
features=df.drop(columns=['Price'])
```

```python
# splitting
x_train,x_test,y_train,y_test=train_test_split(target,features,train_size=0.7,random_state=10)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# which model used
lin=LinearRegression()
lin.fit(x_train,y_train)
```

```python
# training validation
y_pred=lin.predict(x_train)
r2_score(y_train,y_pred)
```

```python
# testing validation
y_pred1=lin.predict(x_test)
r2_score(y_test,y_pred1)
```

```python
lasso=Lasso(alpha=4)
lasso.fit(x_train,y_train)
```

```python
# training validation 
y_pred=lasso.predict(x_train)
r2_score(y_train,y_pred)
```

```python
# testing validation
y_pred1=lasso.predict(x_test)
r2_score(y_test,y_pred1)
```

```python
params=dict(alpha=range(1,5))
params
```

```python
grid_search=GridSearchCV(lasso,params)
grid_search.fit(x_train,y_train)
grid_search.best_params_
```
