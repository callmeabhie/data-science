### TASK 6: Create a list of numbers and,Separate numbers divisible by 3,Separate numbers divisible by 5,Separate numbers divisible by both 3 and 5,Store remaining numbers in another list

```python
abi=[2,45,333,8,5,77,15,90]
div3=[]
div5=[]
div35=[]
num=[]
for i in abi:
    if i%3==0 & i%5==0:
        div35.append(i)
    elif i%3==0:
        div3.append(i)
    elif i%5==0:
        div5.append(i)
    else:
        num.append(i)
print('Divisible by both 3 and 5',div35)
print('Divisible by 3',div3)
print('Divisible by 5',div5)
print('Not in any',num)
```
