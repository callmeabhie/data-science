## Seperate the target from fetures.

```python
target=df[['Price']]
target
```

## Or use this method

```python
features=df.drop(columns=['Price'])
features
```

### import the 1st package for train_test_split

```python
e= sklearn.model_selection import train_test_split
```

### import the second package for model

```python
from sklearn.linear_model import LinearRegression
```

### setting variables and checking rows and columns

```python
x_train,x_test,y_train,y_test= train_test_split(features,target,train_size=0.7,random_state=10)
print(x_train.shape)
print(y_train.shape)
print(y_train.shape)
print(y_test.shape)
```

### model training

```python
lin_reg=LinearRegression()
lin_reg.fit(x_train,y_train)
```

```python
b0=lin_reg.intercept_
b0
```

```python
b1=lin_reg.coef_
b1
```

```python
# training validation
y_predict_tr=lin_reg.predict(x_train)
y_predict_tr
```

```python
# testing validation
y_predict_te=lin_reg.predict(x_test)
y_predict_te
```

### R_Square

```python
from sklearn.metrics import r2_score
```

```python
# training validation
y_predict_tr=lin_reg.predict(x_train)
y_predict_tr
r2_score(y_train,y_predict_tr)
```

```python
# testing validation
y_predict_te=lin_reg.predict(x_test)
y_predict_te
r2_score(y_test,y_predict_te)
```
