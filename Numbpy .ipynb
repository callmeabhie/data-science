```python
# creation of a two dimention matrix using ranndom ranmbers from 1,100.

import numpy as np
two_dim=np.random.randint(1,100,16).reshape(4,4)
two_dim
```

```python
# i have a index=50 ,i need to know the position in matrix.


np.argwhere(two_dim==97)
```

```python
# question: find the position of the index 28?.
```

```python
np.argwhere(two_dim==28)
```

```python
# to insert a new row in position 2 and matrix is [7,12,20,18].

np.insert(two_dim,2,[7,12,20,18])
```

```python
# in python,python dont know where to put the numbers since row and colum have same 2.so introduction of axis
# axis=0------------->X axis.
# axis=1------------->Y axis.
```

```python
#question1:insert [2,34,56,78] in row position=3?.

np.insert(two_dim,3,[2,34,56,78,],axis=0)
```

```python
#question2:insert [3,5,83,99] in column position=4?.

np.insert(two_dim,4,[3,5,83,99],axis=1)
```

```python
# to delete a row matrix, ie; delete a row having position 2.
np.delete(two_dim,2,axis=0)
```

```python
# to delete a column matrix,ie; delete a column having position 3.
np.delete(two_dim,3,axis=1)
```

```python
# to see transpose of a matrix.
np.transpose(two_dim)
```

```python
# to see determinant of a matrix.
np.linalg.det(two_dim)
```

```python
# to see inverse of a matrix.
np.linalg.inv(two_dim)
```
