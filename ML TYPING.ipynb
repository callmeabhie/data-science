```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import f_oneway
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.preprocessing import LabelEncoder,StandardScaler
 
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LinearRegression,Lasso,Ridge,ElasticNet,LogisticRegression
from sklearn.metrics import r2_score,classification_report
```

# 1) LINEAR REGRESIION

```python
df=pd.read_csv('studyhours.csv')
df.head()
```

```python
target=df[['Final_Score']]
features=df.drop(columns='Final_Score')
```

```python
# splitting data
x_train,x_test,y_train,y_test=train_test_split(target,features,train_size=0.7,random_state=10)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# model selection
lin_reg=LinearRegression()
lin_reg.fit(x_train,y_train)
```

```python
# training validation
y_pred=lin_reg.predict(x_train)
r2_score(y_train,y_pred)
```

```python
# testing validation
y_pred1=lin_reg.predict(x_test)
r2_score(y_test,y_pred1)
```

```python
# model selection 
lasso=Lasso()
lasso.fit(x_train,y_train)
```

```python
# training validation
y_pred=lasso.predict(x_train)
r2_score(y_train,y_pred)
```

```python
# testing validation
y_pred1=lasso.predict(x_test)
r2_score(y_test,y_pred1)
```

```python
# model selection 
elastic=ElasticNet()
elastic.fit(x_train,y_train)
```

```python
# training validation 
y_pred=elastic.predict(x_train)
r2_score(y_train,y_pred)
```

```python
# testing validation
y_pred1=elastic.predict(x_test)
r2_score(y_test,y_pred1)
```

```python
# model selection 
ridge=Ridge()
ridge.fit(x_train,y_train)
```

```python
# training validation
y_pred=ridge.predict(x_train)
r2_score(y_train,y_pred)
```

```python
# testing validation
y_pred1=ridge.predict(x_test)
r2_score(y_test,y_pred1)
```

# 2) KNN

```python
df=pd.read_csv('knndataset.csv')
df.head()
```

```python
le=LabelEncoder()
df['Result']=le.fit_transform(df['Result'])
```

```python
target=df[['Result']]
features=df.drop(columns='Result')
```

```python
x_train,x_test,y_train,y_test=train_test_split(target,features,train_size=0.7,random_state=10)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
knn_classi=KNeighborsClassifier(n_neighbors=3)
knn_classi.fit(x_train,y_train)
```

```python
y_pred=knn_classi.predict(x_train)
r2_score(y_train,y_pred)
```

```python
y_pred1=knn_classi.predict(x_test)
r2_score(y_test,y_pred1)
```

# 3) LOGISTIC REGRESSION

```python
df=pd.read_csv('knndataset.csv')
df.head()
```

```python
le=LabelEncoder()
df['Result']=le.fit_transform(df['Result'])
```

```python
target=df[['Result']]
features=df.drop(columns=['Result'])
```

```python
x_train,x_test,y_train,y_test=train_test_split(features,target,train_size=0.7,random_state=10)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
log_reg=LogisticRegression()
log_reg.fit(x_train,y_train)
```

```python
y_pred=log_reg.predict(x_train)
classification_report(y_train,y_pred)
r2_score(y_train,y_pred)
```

```python
y_pred1=log_reg.predict(x_test)
classification_report(y_test,y_pred1)
r2_score(y_test,y_pred1)
```
