# PAIRED TEST

# 7.A study has conducted to determine the effectiveness of a weight base program. The table below shows the before and after weights of 10 subject in the program. A) Is the program effective for reducing weight? B) (use a 5% significant level) construct a 95% confidence interval and determine the margin of error.

```python
#H0--->weight base program is not effective.
#Ha--->weight base program is effective.
```

```python
# GIVEN DATA
n=10
bef=[185,192,206,177,225,168,256,239,199,218]
aft=[169,187,193,176,194,171,228,217,204,195]
# alpha=5%,CI=95%
```

```python
import scipy.stats as stats
import numpy as np
```

```python
# we cant get the difference so convert in array form.
diff=np.array([aft])-np.array([bef])
diff
```

```python
# find mean 
d_bar=np.mean(diff)
d_bar
```

```python
# to find std_bar----> we use function.
# we use ddof=1 because----->This is the sample standard deviation (also called the unbiased estimator), where 
#'𝑛-1' is used instead of 'n' to correct for bias when estimating the population standard deviation from a sample.
sd_bar=np.std(diff,ddof=1)
sd_bar
```

```python
# now find t_stats.
t_stats=d_bar/(sd_bar/np.sqrt(10))
t_stats
```

```python
# now find the t_critical.
t_critical=stats.t.ppf(0.025,9)
t_critical
```

```python
# since the_critical > t_stats in -ve region,Reject H0
# weight base program is effective.
```

```python
# WE DONE THROUGH MANUAL CALCULATION, IT CAN DONE IN USING PACKAGES 
```

```python
stats.ttest_rel(aft,bef)
```

# F-TEST (ANOVA-TEST)

```python
# 9)Anova Test
a= [2,3,7,2,6]
b=[10,8,7,5,10]
c=[10,13,14,13,15]
```

```python
#sample size
n=5
df1=2
df2=12
```

```python
# fstats equation: 
# fstats=(variance_btw_the group/df1)/(variance_in_the group/df2)
```

```python
# variance_btw_the group
mean_a=np.mean(a)
mean_b=np.mean(b)
mean_c=np.mean(c)
print(mean_a,',',mean_b,',',mean_c)
overall_mean=(mean_a+mean_b+mean_c)/3
variance_btw_the_group = n*((overall_mean - mean_a)**2) + n*((overall_mean - mean_b)**2) + n*((overall_mean - mean_c)**2)
variance_btw_the_group                                                                            
```

```python
# variance_within_the_group
var_a=sum((a-mean_a)**2)
var_b=sum((b-mean_b)**2)
var_c=sum((c-mean_c)**2)
variance_within_the_group=var_a+var_b+var_c
variance_within_the_group
```

```python
#know find the fstats
fstats=(variance_btw_the_group/df1)/(variance_within_the_group/df2)
fstats
```

```python
# using package.
```

```python
stats.f_oneway(a,b,c)
```

```python
# find f critical using function.
f_critical=stats.f.ppf(0.95,2,12)
f_critical
```

# NEWPROBLEM

# question: is all this method are equal.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#alpha=5%,CI=95%
df=pd.read_csv('plant_growth1.csv')
df   # -----> # to get first 5 elements
```

```python
# to know the number of rows and columns.
df.shape
```

```python
# H0----> ALL METHOD ARE EQUAL.(ctrl!=trt1!=trt2)
# Ha----> ALL METHOD ARE NOT EQAUL.(ctrl==trt1==trt2)
```

```python
# WE NEED TO SEPERATE THE SAMPLES .AS ,SAMPLE 1,SAMPLE2,SAMPLE3.
```

```python
# to find the types and count
df.treatment_type.value_counts()
```

```python
# we are seperating the types based on type and wieght and making samples.
# its a manual method.
sample1=df[df.treatment_type=='ctrl']['weight']
sample2=df[df.treatment_type=='trt1']['weight']
sample3=df[df.treatment_type=='trt2']['weight']
```

```python
# the function is used to perform a one-way ANOVA test (Analysis Of Variance).
#stats.f_oneway() compares the variance between groups to the variance within groups to calculate an F-statistic and a p-value.
f_stats=stats.f_oneway(sample1,sample2,sample3)
f_stats
```

```python
#treatment_type

# ctrl    10
# trt1    10
# trt2    10
# df1=n-1
# df1=3-1=2
# df2=dfs1+dfs2+dfs3
#    =9+9+9=27
```

```python
# now find the f_critical
f_critical=stats.f.ppf(0.95,2,27)
f_critical
```

```python
# since f_stat > f_critical,reject H0
# ALL METHOD ARE NOT EQAUL.(ctrl==trt1==trt2)
```
